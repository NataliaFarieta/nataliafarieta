**Identificación de valores atípicos**

Se detectan valores atípicos utilizando el método del Rango Intercuartílico (IQR), con el fin de evaluar su impacto potencial en los modelos de regresión. De los datos obtenidos se puede indicar que:

+ **RM** Para el numero promedio de habitaciones se presenta valores atípicos de 4,5%, correspondiente a viviendas muy grandes en zonas de alto costo de poder adquisitivo.
+ **LSTAT** Para el % de población de bajo estatus, tiene pocos outliers 1,4%, que probablemente se traten de barrios muy marginales.
+ **PTRATIO** La relación entre alumno-profesor también posee pocos valores atípicos solo un 2,7%, representada por colegios con gran cantidad de estudiantes por profesor.
+ **MEDV** En cuanto al valor medio de las viviendas, posee una una cantidad importante de outliers.

En general las variables analizadas presentan un bajo porcentaje de valores atípicos según el criterio IQR (≤ 4.5%), lo que sugiere que la base es estable y e puede usar para la regresión, sin necesidad de imputar datos.


```python
# Identificación de valores atípicos (outliers) mediante el método del Rango Intercuartílico (IQR)

# Función que identifica outliers en una variable numérica usando el criterio IQR
def iqr_outlier_mask(s: pd.Series) -> pd.Series:
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    return (s < lower) | (s > upper)

# Cálculo y reporte del número y porcentaje de outliers por variable
for col in df.columns:
    mask = iqr_outlier_mask(df[col])
    print(f'Outliers por IQR en {col}: {mask.sum()} ({mask.mean():.1%})')
```

##**Preparación para el modelado**

Se importan las librerías necesarias para la construcción de las regresiones lineales y multiples, para luego dividir el dataset en conjuntos de entrenamiento y prueba con el fin de una evaluación adecuada del desempeño de los modelos.

```python
# Importación de librerías de scikit-learn para el modelado

# train_test_split: divide los datos en conjuntos de entrenamiento y prueba
from sklearn.model_selection import train_test_split

# LinearRegression: modelo de regresión lineal
from sklearn.linear_model import LinearRegression

# Métricas para evaluar el desempeño del modelo
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

#**Modelo de Regresión Lineal Simple**

A continuación, se ajusta el modelo de regresión lineal simple, utilizando cada variable predictora de forma individual para explicar la variable objetivo MEDV. Para cada predictor, el modelo es entrenado con el conjunto de datos de entrenamiento y evaluado con el conjunto de prueba, este análisis permite comprender qué variables tienen mayor influencia individual sobre MEDV, aunque su capacidad predictiva es limitada ya que no considera todas las variables en conjunto.

El desempeño de cada modelo se mide mediante el coeficiente de determinación R², así como métricas de error (MAE y RMSE), con el propósito de comparar la capacidad explicativa de cada variable por separado. A partir de esta comparación, se identifica la variable predictora que presenta el mayor valor de R², considerándose como el mejor modelo simple para explicar el valor medio de las viviendas.

Se aplica la siguiente fórmula siendo **X:** variable predictora a estudiar
image.png


```python
# Definición de variables predictoras (X) y variable objetivo (y)
X = df[features]
y = df[target]

# División de los datos en conjunto de entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
```

```python
# Función para evaluar el desempeño del modelo de regresión

# Calcula métricas de evaluación: R², MAE y RMSE
def eval_model(y_true, y_pred):
    return {
        "R2": r2_score(y_true, y_pred),
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred))
    }
```

```python
metricas_modelo = []

variables = X.columns
n = len(variables)

fig, axes = plt.subplots(1, n, figsize=(5*n, 4))

for i, col in enumerate(variables):
    lr = LinearRegression()
    lr.fit(X_train[[col]], y_train)
    y_pred = lr.predict(X_test[[col]])

    # Métricas
    r2 = r2_score(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))

    metricas_modelo.append({
        "Variable": col,
        "R2": r2,
        "MAE": mae,
        "RMSE": rmse
    })

    # Gráfico en subplot
    axes[i].scatter(X_test[col], y_test, alpha=0.6, label='Real')
    axes[i].scatter(X_test[col], y_pred, alpha=0.6, label='Predicho')
    axes[i].set_title(f'MEDV vs {col}\nR²={r2:.3f}', fontsize=10)
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('MEDV')
    axes[i].legend()
    axes[i].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

**Evaluación del modelo**

Para la correcta evaluacion del desempeño de los modelos, se define una función de evaluación que permite medir métricas como R², MAE y RMSE. Que permite inferir que:

+ **RM** Tiene una relacion positiva moderada con MEDV con un R2 de 0,3919 indicaque a mayor numero promedio de haitaciones es mayor el precio de la vivienda.
+ **LSTAT** Es la variable que tiene mayor R2 0.5683, ya que a mayor porcentaje de poblacion de bajo estatus, menor es el valor de la vivienda.
+ **PTRATIO** No es una variable muy predictiva,tiene un R2 de 0.1761, con una dispesion alta, por ende no tiene un alta influencia en el valor de la vivienda .

En la comparacion de todos los R2 obtenidos se concluye que apesar de que LSTAT Y RM tienen R2 relativamente algo,nunguna variable por si sola no justifica claramente el valor de la vivienda, es necesario aplicar la regresion multiple para mejor la capacidad predictiva, ya que tiene en cuenta todas las variables simultaneamente.

```python
df_metricas = pd.DataFrame(metricas_modelo)
df_metricas.sort_values("R2",ascending=False)
```

```python
plt.figure()
plt.bar(df_metricas["Variable"], df_metricas["R2"])
plt.xlabel("R2")
plt.title("Comparaciones de modelos de regresión lineal simple")
#plt.gca().invert_yaxis()  # Para que la primera variable aparezca arriba
plt.show()
```
